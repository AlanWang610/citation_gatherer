```python
from habanero import Crossref

# Initialize CrossRef client
cr = Crossref()

# Search parameters
author = "Bernanke Ben S"
year = "1989"
journal = "A.E.R."
first_page = "257"

# Perform search query
query = f"{author} {year} {journal} {first_page}"
results = cr.works(query=query)

# Print the first matching result
if results['message']['items']:
    first_result = results['message']['items'][0]
    title = first_result.get("title", ["No title found"])[0]
    doi = first_result.get("DOI", "No DOI found")
    print(f"Title: {title}")
    print(f"DOI: {doi}")
else:
    print("No results found.")
```

```python
from habanero import Crossref

# Initialize CrossRef client
cr = Crossref()

# Search parameters
author = "Calomiris Charles W"
volume = "81"
journal = "A.E.R."
first_page = "497"

# Perform search query
query = f"{author} {volume} {journal} {first_page}"
results = cr.works(query=query)

# Print the first five matching results
if results['message']['items']:
    for i, result in enumerate(results['message']['items'][:5]):
        title = result.get("title", ["No title found"])[0]
        doi = result.get("DOI", "No DOI found")
        print(f"\nResult {i+1}:")
        print(f"Title: {title}")
        print(f"DOI: {doi}")
else:
    print("No results found.")
```
